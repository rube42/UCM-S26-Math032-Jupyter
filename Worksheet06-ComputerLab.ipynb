<h4> Drawing random numbers from a discrete distribution.</h4>

In problem 1 of the worksheet, you considered a dicrete distribution for which the probability mass function was specified. In class, we  introduced a method for generating realizations of discrete random variables using the cumulative distribution function. Below is a code snipped that implements the six-sided dice:

```python
# Import the necessary package
from numpy import random

#First generate a random number between 0 and 1
r = random.rand()

#Next we use the cumulative distribution function to figure out what number on the die this corresponds to: 
if r<=1/6:     #F(1) = 1/6
    die = 1
elif r<=2/6:   #F(2) = 2/6
    die = 2
elif r<=3/6:   #F(3) = 3/6
    die = 3
elif r<=4/6:   #F(4) = 4/6
    die = 4
elif r<=5/6:   #F(5) = 5/6
    die = 5
else:
    die = 6    #F(5) = 1

print("The die is",die)

```

To check if if this works properly, one can generating many dice rolls and see if the probability of getting a 6 is close to 1/6:


```python
#How many samples to draw
nSample = 10000
#Counts the number of sixes
n6 = 0

for _ in range(nSample):
    #Generates the die roll
    r = random.rand()
    if r<1/6:
        die = 1
    elif r<2/6:
        die = 2
    elif r<3/6:
        die = 3
    elif r<4/6:
        die = 4
    elif r<5/6:
        die = 5
    else:
        die = 6
        
    #Checks if it is six:
    if die == 6:
        n6 += 1
        
print("P[X=6] = ",n6/nSample)
```

We see that this is close to 1/6=0.16667. 

<h4> Computing expectation values using simulations</h4>

To check the worksheet we also need to compute expectation values using simulations. Remember that one definition of the expectation value $E[X]$ of a random variable $X$ is the mean observed values after repeating a random experimet indefinitely. To illustrate how expectation values can be approximated using simulations, consider $X\sim N(\mu=5, \sigma^2=1)$. To estimate $\mu=E[X]$ and $Var(X)=E[X^2]-\mu^2$ using a simulation, we compute the mean value of $X$ and $X^2$ after generating 10,000 realizations of X:

```python
from random import gauss

nSample = 10000
#Sum used to compute E[X]
xSum  = 0
#Sum used to compute E[X^2]
x2Sum = 0
for _ in range(nSample):
    x = gauss(5,1)
    xSum  += x
    x2Sum += x**2
    
EX  = xSum / nSample
EX2 = x2Sum / nSample
Var = EX2 - EX**2
print("E[X]   = ",EX)
print("E[X^2] = ",EX2)
print("Var[X] = ",Var)
```

We see that the $X$ has the true mean and variance ($\mu=5$, $\sigma^2=1$).
